```python
import pandas as pd
```

```python
data = pd.read_csv("/kaggle/input/datasets/aiexplorer77/from-fuel-to-electric-the-ev-transition-dataset/global_ev_adoption_behavior_2026.csv")
```

```python
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
# Data preprocessing 

from sklearn.impute import SimpleImputer

num_imputer = SimpleImputer(strategy='median')

data[['charging_station_accessibility', 'ev_knowledge_score']] = num_imputer.fit_transform(
    data[['charging_station_accessibility', 'ev_knowledge_score']]
)

freq_imputer = SimpleImputer(strategy='most_frequent')
data[['education_level']] = freq_imputer.fit_transform(data[['education_level']])
```

```python
data.info()
```

```python
# Data Encoding 

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()
data["ev_adoption_likelihood"] = le.fit_transform(data["ev_adoption_likelihood"])

col = ["city_type","current_vehicle_type"]
data = pd.get_dummies(data, columns=col, drop_first=True, dtype=int)
```

```python
data.head(10)
```

```python
X = data.drop(columns = ["education_level","ev_adoption_likelihood", "monthly_energy_consumption_kwh", "monthly_charging_cost"], axis = 1)
Y = data["ev_adoption_likelihood"]
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X,Y,
    test_size=0.4,
    random_state=42
)
```

```python
X_train.head()
```

```python
Y_train.head()
```

```python
# Data Scaling

from sklearn.preprocessing import StandardScaler

scale = StandardScaler()

X_train = scale.fit_transform(X_train)
X_test = scale.transform(X_test)
```

# SVM Classifier

```python
# Default kernel "RBF"

from sklearn.svm import SVC

svc = SVC()
svc.fit(X_train, Y_train)
```

```python
Y_pred = svc.predict(X_test)
Y_pred_t = svc.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score

print(f"Accuracy for train data: {accuracy_score(Y_train,Y_pred_t)*100} %")
print(f"Accuracy for test data: {accuracy_score(Y_test,Y_pred)*100} %")
```

## Different Kernel Methods

```python
svc = SVC(kernel = "linear")
svc.fit(X_train, Y_train)

Y_pred = svc.predict(X_test)
Y_pred_t = svc.predict(X_train)

print(f"Accuracy for train data: {accuracy_score(Y_train,Y_pred_t)*100} %")
print(f"Accuracy for test data: {accuracy_score(Y_test,Y_pred)*100} %")
```

```python
svc = SVC(kernel = "poly")
svc.fit(X_train, Y_train)

Y_pred = svc.predict(X_test)
Y_pred_t = svc.predict(X_train)

print(f"Accuracy for train data: {accuracy_score(Y_train,Y_pred_t)*100} %")
print(f"Accuracy for test data: {accuracy_score(Y_test,Y_pred)*100} %")
```
