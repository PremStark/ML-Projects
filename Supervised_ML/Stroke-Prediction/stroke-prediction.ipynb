```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
data = pd.read_csv("/kaggle/input/datasets/fedesoriano/stroke-prediction-dataset/healthcare-dataset-stroke-data.csv")
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data.info()
```

```python
# Data Preprocessing

from sklearn.impute import SimpleImputer

imp_mean = SimpleImputer(strategy="mean")
data[["bmi"]] = imp_mean.fit_transform(data[["bmi"]])
```

```python
data.head(10)
```

```python
data.info()
```

```python
# Data Encoding

col = ["gender","ever_married","work_type", "Residence_type", "smoking_status"]
data = pd.get_dummies(data, columns=col, drop_first=True, dtype=int)
```

```python
data.head()
```

```python
X = data.drop("stroke", axis=1)
Y = data["stroke"]
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(
    X,Y,
    test_size=0.3,
    random_state=42
)
```

# Random Forest Classifier

```python
from sklearn.ensemble import RandomForestClassifier

rfc = RandomForestClassifier(
    n_estimators=1000,
    max_depth = 4,
    random_state = 42
)
```

```python
rfc.fit(X_train, Y_train)
```

```python
Y_pred = rfc.predict(X_test)
Y_pred_t = rfc.predict(X_train)
```

```python
print(f"Accuracy score for train: {accuracy_score(Y_train,Y_pred_t)*100} %")
print(f"Accuracy score for test: {accuracy_score(Y_test,Y_pred)*100} %")
```
