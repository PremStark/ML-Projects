```python
import pandas as pd
```

```python
loan_data = pd.read_csv("/kaggle/input/datasets/premraj25/loan-data/loan_approval_data.csv")
```

```python
loan_data.head()
```

```python
loan_data.info()
```

```python
loan_data.isnull().sum()
```

```python
loan_data = loan_data.drop(columns=["Applicant_ID"],axis=1)
```

```python
loan_data.head()
```

```python
loan_data.isnull().sum()
```

```python
# Data Preprocessing 

from sklearn.impute import SimpleImputer

# For numerical columns

imp_mean = SimpleImputer(strategy="mean")

features_val = ["Applicant_Income","Coapplicant_Income","Age", "Dependents", "Credit_Score", "Existing_Loans", "DTI_Ratio", "Savings", "Collateral_Value", "Loan_Amount", "Loan_Term"]

for col in features_val:
    loan_data[[col]] = imp_mean.fit_transform(loan_data[[col]])

# For Category columns

imp_freq = SimpleImputer(strategy="most_frequent")

features_cat = ["Employment_Status", "Marital_Status", "Loan_Purpose", "Property_Area", "Education_Level", "Employer_Category", "Gender","Loan_Approved"]

for col in features_cat:
    loan_data[[col]] = imp_freq.fit_transform(loan_data[[col]])
```

```python
loan_data.isnull().sum()
```

```python
loan_data.head()
```

```python
# Encoding for Category Columns

from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()

feature_encod = ["Employment_Status", "Marital_Status", "Loan_Purpose", "Property_Area", "Education_Level", "Employer_Category", "Gender","Loan_Approved"]

for col in feature_encod:
    loan_data[col] = le.fit_transform(loan_data[col])
```

```python
loan_data.head(10)
```

```python
# Building DT model

from sklearn.model_selection import train_test_split

X = loan_data.drop("Loan_Approved", axis = 1)
Y = loan_data["Loan_Approved"]

X.head()
```

```python
Y.head()
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y,
    test_size = 0.3,                                                
    random_state = 42
)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(X_train, Y_train)
```

```python
acc = model.score(X_test, Y_test)
print(f"Accuracy of ML Model for Loan approval: {acc*100} %")
```
